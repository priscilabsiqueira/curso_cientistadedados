**Efetividade de Duas Estratégias de Ensino**

Imagine que uma escola esteja avaliando a eficácia de duas estratégias de ensino de matemática para alunos do ensino médio. Eles querem determinar se há uma diferença significativa no desempenho médio dos alunos entre as duas estratégias.

# **Hipóteses:**

* Hipótese nula (H0): A média das notas dos alunos na estratégia A é igual à média das notas dos alunos na estratégia B.
* Hipótese alternativa (H1): A média das notas na Estratégia B é maior do que a média das notas na Estratégia A.

# **Dados:**

* Amostra da Estratégia A: Notas de 50 alunos que receberam a Estratégia A.
* Amostra da Estratégia B: Notas de 50 alunos que receberam a Estratégia B.

Usaremos um teste Z para comparar as médias das notas entre as duas amostras.

Se o p-valor do teste Z for menor que um nível de significância pré-determinado (por exemplo, α = 0.05), rejeitamos a hipótese nula e concluímos que há uma diferença significativa nas médias das notas entre as duas estratégias de ensino.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

Os dados são criados a seguir:

```python
# Definindo médias e desvios padrão para as notas nas duas estratégias
media_estrategia_A = 70
desvio_padrao_estrategia_A = 10

media_estrategia_B = 75
desvio_padrao_estrategia_B = 12

# Gerando as amostras de notas para cada estratégia de ensino da nossa base
np.random.seed(0)  # Para garantir a reprodutibilidade dos resultados
amostra_estrategia_A = np.random.normal(loc=media_estrategia_A, scale=desvio_padrao_estrategia_A, size=50)
amostra_estrategia_B = np.random.normal(loc=media_estrategia_B, scale=desvio_padrao_estrategia_B, size=50)

print("Notas da Estratégia A:", amostra_estrategia_A[:5])
print("Notas da Estratégia B:", amostra_estrategia_B[:5])
```

# 1) De acordo com as informações analisadas o nosso teste é unilateral á direita, esquerda ou bicaudal? Justifique.




# Sua justificativa aqui

Bicaudal pois a análise quer saber se há uma diferença entre estratégia A e B.

# 2) Calcule as médias para as duas amostragens e as variâncias. Quais insights você pode retirar comparando os dados?

```python
#Seu código aqui

media_amostra_A = np.mean(amostra_estrategia_A)
media_amostra_B = np.mean(amostra_estrategia_B)

# Calculando as variâncias das amostras (usando ddof=1 para variância amostral não viesada)
variancia_amostra_A = np.var(amostra_estrategia_A, ddof=1)
variancia_amostra_B = np.var(amostra_estrategia_B, ddof=1)

print(f"Média da Amostra A: {media_amostra_A:.2f}")
print(f"Variância da Amostra A: {variancia_amostra_A:.2f}")
print(f"Média da Amostra B: {media_amostra_B:.2f}")
print(f"Variância da Amostra B: {variancia_amostra_B:.2f}")
```

# Média da Amostra B é maior que a média da Amostra A.

# 3) Imprima os resultados da estatística do teste Z, p value e indique se rejeitaremos ou não a hipótese nula. Justifique sua resposta.

```python
#Seu código aqui
media_amostra_A = np.mean(amostra_estrategia_A)
media_amostra_B = np.mean(amostra_estrategia_B)
n_A = len(amostra_estrategia_A)
n_B = len(amostra_estrategia_B)

# Calcular a estatística de teste Z
# Usamos (media_amostra_A - media_amostra_B) para a diferença observada.
# No denominador, usamos os desvios padrão POPULACIONAIS (ao quadrado para variância)
# divididos pelos tamanhos das respectivas amostras.
z_statistic = (media_amostra_A - media_amostra_B) / \
              ((desvio_padrao_estrategia_A**2 / n_A) + (desvio_padrao_estrategia_B**2 / n_B))**0.5

# Calcular o p-valor (teste bicaudal)
# stats.norm.cdf(abs(z_statistic)): Calcula a área sob a curva normal padrão até o valor absoluto da estatística Z.
# 2 * ...: Multiplicamos por 2 porque é um teste bicaudal (queremos a probabilidade nas duas caudas).
p_valor = 2 * (1 - stats.norm.cdf(abs(z_statistic)))


# Imprimir os resultados
print(f"Média da Amostra A: {media_amostra_A:.2f}")
print(f"Média da Amostra B: {media_amostra_B:.2f}")
print("-" * 30)
print(f"Estatística de teste Z: {z_statistic:.4f}")
print(f"P-valor: {p_valor:.4f}")

# Comparar o p-valor com o nível de significância
nivel_significancia = 0.05
print(f"Nível de Significância (alpha): {nivel_significancia}")

if p_valor < nivel_significancia:
    print("\nDecisão: Rejeitamos a hipótese nula.")
    print("Justificativa: Há evidências estatisticamente significativas para concluir que existe uma diferença no desempenho médio dos alunos entre as duas estratégias de ensino.")
else:
    print("\nDecisão: Não rejeitamos a hipótese nula.")
    print("Justificativa: Não há evidências estatisticamente significativas para concluir que existe uma diferença no desempenho médio dos alunos entre as duas estratégias de ensino.")
```

# 4) Para finalizar monte o gráfico da distribuição da estatística do nosso teste Z e explique o que pode ser observado através dele.

```python
#Seu código aqui

# --- Geração do gráfico ---
plt.figure(figsize=(10, 6))
# Gerar pontos para a curva normal padrão
x = np.linspace(-4, 4, 1000)
plt.plot(x, stats.norm.pdf(x), color='blue', label='Distribuição Normal Padrão (H0)')

# Adicionar a estatística Z
plt.axvline(z_statistic, color='red', linestyle='--', label=f'Z Estatística = {z_statistic:.2f}')

# Adicionar o nível de significância (pontos críticos para um teste bicaudal)
# Para alfa = 0.05 bicaudal, os valores críticos são aproximadamente -1.96 e 1.96
z_critical_positive = stats.norm.ppf(1 - nivel_significancia / 2)
z_critical_negative = stats.norm.ppf(nivel_significancia / 2)

plt.axvline(z_critical_positive, color='green', linestyle=':', label=f'Z Crítico (+{z_critical_positive:.2f})')
plt.axvline(z_critical_negative, color='green', linestyle=':', label=f'Z Crítico ({z_critical_negative:.2f})')


# Sombrear as regiões de rejeição (bicaudal)
# Região da cauda esquerda
x_rejection_left = x[x < z_critical_negative]
plt.fill_between(x_rejection_left, 0, stats.norm.pdf(x_rejection_left), color='gray', alpha=0.3, label='Região de Rejeição (α/2)')

# Região da cauda direita
x_rejection_right = x[x > z_critical_positive]
plt.fill_between(x_rejection_right, 0, stats.norm.pdf(x_rejection_right), color='gray', alpha=0.3)


plt.title('Distribuição da Estatística Z com Regiões de Rejeição (Teste Bicaudal)')
plt.xlabel('Estatística Z')
plt.ylabel('Densidade de Probabilidade')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()

# Resumo da decisão
print(f"Estatística Z: {z_statistic:.4f}")
print(f"P-valor: {p_valor:.4f}")
print(f"Nível de Significância (alpha): {nivel_significancia}")

if p_valor < nivel_significancia:
    print("\nDecisão: Rejeitamos a hipótese nula.")
    print("Justificativa: O p-valor é menor que o nível de significância, indicando evidência significativa de diferença.")
else:
    print("\nDecisão: Não rejeitamos a hipótese nula.")
    print("Justificativa: O p-valor é maior ou igual ao nível de significância, não havendo evidência suficiente para concluir uma diferença.")
```
